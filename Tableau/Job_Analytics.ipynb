```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

# The original error was due to passing None as the path for a PANDAS adapter.
# When using the PANDAS adapter, a specific file within the dataset must be specified.

# Load the latest version of a specific file from the dataset
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "ravindrasinghrana/job-description-dataset",
  # Specify the name of a CSV file within the dataset.
  # This is a common filename, but you might need to adjust it
  # if the actual file has a different name (e.g., 'data.csv', 'jobs.csv').
  "job_descriptions.csv",
  # Provide any additional arguments like
  # sql_query or pandas_kwargs. See the
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
)

print("First 5 records:")
display(df.head())
```

```python
intern_count = (df['Work Type'] == 'Intern').sum()
print(f"Number of 'Intern' positions: {intern_count}")

job_portal_count = (df['Job Portal'] == 'Indeed').sum()
print(f"Number of 'Indeed' in job portal: {job_portal_count}")

job_portal_count2 = (df['Job Portal'] == 'LinkedIn').sum()
print(f"Number of 'Linkedin' in qualifications: {job_portal_count2}")

qualification_count = (df['Qualifications'] == 'B.Tech').sum()
print(f"Number of 'B.tech' in qualifications: {qualification_count}")

qualification_count2 = (df['Qualifications'] == 'M.Tech').sum()
print(f"Number of 'M.Tech' in qualifications: {qualification_count2}")

qualification_count3 = (df['Qualifications'] == 'PhD').sum()
print(f"Number of 'Phd' in qualifications: {qualification_count3}")

jobtitle_count4 = (df['Job Title'] == 'Mechanical Engineer').sum()
print(f"Number of 'Mechanical Engineeri' in job title: {jobtitle_count4}")

jobtitle_count5 = (df['Job Title'] == 'Data Scientist').sum()
print(f"Number of 'Data Scientist' in job title: {jobtitle_count5}")

jobtitle_count6 = (df['Job Title'] == 'Data Engineer').sum()
print(f"Number of 'Data Engineer' in job title: {jobtitle_count6}")
```

# Removing unnecessary columns

```python
df.drop(columns=['Contact','Job Description','Benefits','skills','Responsibilities'], inplace=True)
```

```python
df = df.rename(columns={'Job Id':'job_id',
                        'Salary Range':'salary_range',
                        'Work Type':'work_type',
                        'Company Size':'company_size',
                        'Job Posting Date':'job_post_date',
                        'Contact Person':'contact',
                        'Job Title':'job_title',
                        'Job Portal':'job_portal',})
df
```

# General Formating for the dataset

```python
df[['min_salary', 'max_salary']] = df['salary_range'].str.split('-', expand=True)
df['min_salary'] = df['min_salary'].str.replace('$', '').str.replace('K', '').astype(int)
df['max_salary'] = df['max_salary'].str.replace('$', '').str.replace('K', '').astype(int)
df['min_salary'] = df['min_salary'] * 1000
df['max_salary'] = df['max_salary'] * 1000
df
```

```python
df.to_csv('job_descriptions_clean.csv', index=False)
print("DataFrame saved to 'job_descriptions_clean.csv'")
```
